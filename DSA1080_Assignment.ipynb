## Part D: Shopping Basket Calculator

```python
num_products = int(input("How many products do you want to buy: "))
print("Number of products:", num_products)
total = 0
expensive_count = 0

for b in range(num_products):
    price = float(input("Enter the price of product " + str(b + 1) + ": "))
    total = total + price
    if price > 1000:
        expensive_count = expensive_count + 1

if total >= 10000:
    discount_percent = 15
elif total >= 5000:
    discount_percent = 10
elif total >= 2000:
    discount_percent = 5
else:
    discount_percent = 0

discount_amount = total * discount_percent / 100
final_amount = total - discount_amount

print("Total:", total)
print("Discount amount:", discount_amount)
print("Final payable:", final_amount)
print("Products over KSh 1,000:", expensive_count)
```

## Part E: Simple ATM Simulator


```python
balance = 10000

while True:
    print("\n===== SIMPLE ATM =====")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        print("Your balance is: KSh", balance)

    elif choice == "2":
        amount = float(input("Enter amount to deposit: "))
        if amount <= 0:
            print("Invalid amount. Deposit must be greater than 0.")
        else:
            balance = balance + amount
            print("Deposit successful. New balance: KSh", balance)

    elif choice == "3":
        amount = float(input("Enter amount to withdraw: "))
        if amount <= 0:
            print("Invalid amount. Withdrawal must be greater than 0.")
        elif amount > balance:
            print("Insufficient funds. Your balance is KSh", balance)
        else:
            balance = balance - amount
            print("Withdrawal successful. New balance: KSh", balance)

    elif choice == "4":
        print("Thank you for using the ATM. Goodbye!")
        break

    else:
        print("Invalid choice. Please enter 1, 2, 3 or 4.")
```

```python
# Part F - Break and Continue
count = 0

while True:
    number = float(input("Enter a number (0 to stop): "))

    if number == 0:
        break

    if number < 0:
        print("Negative number skipped.")
        continue

    print("The square of", number, "is", number * number)
    count = count + 1

print("Positive numbers processed:", count)
```

```python
# Part G - Savings Tracker
saved = 189
weeks = 0
goal = float(input("Enter your savings goal (KSh): "))

if goal <= saved:
    print("You have already reached this goal!")
else:
    while saved < goal:
        amount = float(input("Enter amount saved this week (0 to stop): "))

        if amount == 0:
            print("Tracker stopped.")
            break

        if amount < 0:
            print("Invalid amount. Please enter a positive number.")
            continue

        saved = saved + amount
        weeks = weeks + 1
        print("Week", weeks, "- Total saved: KSh", saved)

        if saved >= goal:
            print("You've hit your goal!")
        elif saved >= goal * 0.75:
            print("Almost there, keep going!")
        elif saved >= goal / 2:
            print("You're more than halfway there!")
        else:
            print("Good start, keep saving!")

    if saved >= goal:
        print("Goal reached in", weeks, "weeks!")
    else:
        print("You saved KSh", saved, "in", weeks, "weeks. KSh", goal - saved, "short of your goal.")
```

- The starting balance is 189 as those are the last three digits of my student ID

- This program uses (while, if, elif, else, break, continue, and input.) 

1. The first part enables you to input your savings goal.

2. If the amount that you have saved is greater than or equal to your goal, the program will print "You have already reached this goal!"
But if not it will enter the - while loop that states if your the amount you have is less than your goal it will keep the loop on. If you input 0 it will break the while loop. 

3. If you input a negative number it will ask you to input a positive number

4. Then there are milestones to show you where you are at. 
    If you are above half it will print "You're more than halfway there" etc. 
    If you are above 75% it will print "Almost there, keep going!".
    If the amount you have is equal to the goal or greater than the goal it will print "You've hit your goal!".
    Otherwise it will print "Good start, keep saving!"

5. When finally you reach your goal it will count how many times you saved per week to show you how many weeks it took you to get there. 
It will also print how much you have saved and how long it has taken you to get there.
